```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x))
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(0, 10, 100)
fig = plt.figure()
plt.plot(x, np.sin(x), "o")
plt.plot(x, np.cos(x), "r--")
fig.savefig("my_figure.png")
```

```python
x=np.linspace(0, 10, 100)
#matlab-style interface
plt.figure()
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x))
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x))
plt.subplot(2, 2, 3 )
plt.plot(x, np.sin(x))
plt.subplot(2, 2, 4)
plt.plot(x, np.cos(x))
plt.show()
```

```python
import numpy as np

x = np.linspace(0, 10, 100)
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

fig, ax = plt.subplots(2, 2)
ax[0, 0].plot(x, np.sin(x))
ax[0, 1].plot(x, np.cos(x))
ax[1, 0].plot(x, np.tan(x))
ax[1, 1].plot(x, np.exp(x))
plt.style.use('dark_background')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x-0), color='blue', alpha=0.5)
plt.plot(x, np.sin(x-1),color='black', alpha=0.5)
plt.plot(x, np.sin(x-2),color='red', alpha=0.5)
plt.plot(x, np.sin(x-3),color='green', alpha=0.5)
plt.plot(x, np.sin(x-4),color='orange', alpha=0.5)
plt.plot(x, np.sin(x-5),color='purple', alpha=0.5)
plt.figtext(0.5, 0.01, "This is a figure-level text", ha="center", fontsize=12)

```
